Given a dataframe containing records of student and their marks, and the cutoff marks k

Return the name of students who scored marks greater than or equal to the cuttof i.e. marks >= k

A DataFrame containing marks records
k (cutoff marks)
Output Format:
A Series


```python
import pandas as pd

data = {
    'name': ['Luffy', 'Chopper', 'Zoro', 'Sanji'],
    'marks': [15, 89, 33, 32]
}

df = pd.DataFrame(data)


def filter(df, k):
    '''
    INPUT:
        df -> dataframe
        k -> cutoff (int)

    OUTPUT:
        result -> dataframe
    '''

    ### STEP 1 : Create mask
    mask = df['marks'] >= k

    ### STEP 2: Use mask to filter rows
    result = df[mask]['name']

    return result
```

```python
filter(df,30)
```

Given a dataframe containing student's marks for multiple subjects,

Return a dataframe which contains records of subject-wise highest marks along with their roll number.

Input Format:

A DataFrame
Output Format:

A DataFrame

```python
import pandas as pd

def get_marks(df):
    '''
    INPUT: df -> dataframe

    OUTPUT: result -> dataframe
    '''

    ### STEP 1: group the df using subject and get max marks
    max_marks = df.groupby('subject')['marks'].max()

    ### STEP 2: reset the index
    max_marks = max_marks.reset_index()

    ### STEP 3: merge with original df to get roll id
    result = pd.merge(df, max_marks, on=['subject','marks'])

    return result
```
