#Slicing is for continuous subpart os an array

```python
m1=np.arange(10,20)
m1
```

```python
m1[:4]  # starts at 0 but ends at 3 - in slice start value is iclusive but end value is n-1 
```

```python
m1[-5:-1] #start at -5 pos and goes till -1-1 (-2)
```

```python
m1[-5:-1:-1]  # it will never reach to -1   ml[n,m]   n, n+1 ....m-1
```
